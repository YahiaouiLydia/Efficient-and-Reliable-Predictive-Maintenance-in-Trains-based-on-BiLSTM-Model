# Data Exploration

In this notebook, we will load and explore the raw datasets `MetroPT3` and `MetroPT`. The goal is to understand their structure, inspect some initial rows, identify any missing values, and perform basic visualizations for a better understanding of the data.


## Load the datasets

```python
from IPython.display import display, HTML
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import yaml

# Set plot style for better visuals
sns.set(style="whitegrid")

# Load data using the paths from the config file
project_root = Path().resolve().parent
config_path = project_root / "config" / "config.yaml"
with open(config_path, "r") as f:
    config = yaml.safe_load(f)

# Build full paths
path_metropt3 = project_root / config["paths"]["raw"]["metropt3"]
path_metropt = project_root / config["paths"]["raw"]["metropt"]

# Load the datasets
df_metropt3 = pd.read_csv(path_metropt3)
df_metropt = pd.read_csv(path_metropt)

# Displaying the dataset previews with styled titles for better visibility in dark mode
display(HTML('<h1 style="color:#00BFFF; font-size:20px; font-weight:bold; background-color:#333333; padding:10px;">MetroPT3 Dataset Preview:</h1>'))
print(df_metropt3.head())

display(HTML('<h1 style="color:#00BFFF; font-size:20px; font-weight:bold; background-color:#333333; padding:10px;">MetroPT Dataset Preview:</h1>'))
print(df_metropt.head())
```

### Dataset 1: MetroPT3

Let's begin by inspecting the structure of the `MetroPT3` dataset. We will check the columns, data types, and look for any missing values. This will help us better understand the dataset and identify any necessary preprocessing steps.


```python
display(HTML('<h1 style="color:#00BFFF; font-size:20px; font-weight:bold; background-color:#333333; padding:10px;">Dataset Information (Columns and Data Types):</h1>'))
print(df_metropt3.info())

display(HTML('<h1 style="color:#00BFFF; font-size:20px; font-weight:bold; background-color:#333333; padding:10px;">Missing Values per Column:</h1>'))
print(df_metropt3.isnull().sum())

display(HTML('<h1 style="color:#00BFFF; font-size:20px; font-weight:bold; background-color:#333333; padding:10px;">Summary Statistics (for numerical columns):</h1>'))
print(df_metropt3.describe())
```

### Dataset 2: MetroPT

Now, let's move on to the `MetroPT` dataset. We will follow the same approach: inspect the structure, check for missing values, and analyze the dataset’s basic statistics.


```python
# Inspect the columns and data types of the MetroPT dataset
display(HTML('<h1 style="color:#00BFFF; font-size:20px; font-weight:bold; background-color:#333333; padding:10px;">Dataset Information (Columns and Data Types):</h1>'))
print(df_metropt.info())

# Check for missing values in the MetroPT dataset
display(HTML('<h1 style="color:#00BFFF; font-size:20px; font-weight:bold; background-color:#333333; padding:10px;">Missing Values per Column:</h1>'))
print(df_metropt.isnull().sum())

# Show basic summary statistics for the MetroPT dataset
display(HTML('<h1 style="color:#00BFFF; font-size:20px; font-weight:bold; background-color:#333333; padding:10px;">Summary Statistics (for numerical columns):</h1>'))
print(df_metropt.describe())
```

### Data Visualization

Next, we'll create visualizations to further explore the relationships between features. 

#### Data Visualization for MetroPT3 Dataset

```python
# Correlation heatmap for the MetroPT3 dataset
plt.figure(figsize=(12, 8))
corr_metropt3 = df_metropt3.corr()
sns.heatmap(corr_metropt3, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title("Correlation Heatmap - MetroPT3 Dataset")
plt.show()

# Distribution of key numeric features in MetroPT3
df_metropt3.select_dtypes(include=['float64', 'int64']).hist(figsize=(15, 10), bins=20)
plt.suptitle("Distribution of Numeric Features - MetroPT3 Dataset")
plt.show()
```

#### Data Visualization for MetroPT Dataset

```python
# Correlation heatmap for the MetroPT dataset
plt.figure(figsize=(12, 8))
corr_metropt = df_metropt.corr()
sns.heatmap(corr_metropt, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title("Correlation Heatmap - MetroPT Dataset")
plt.show()

# Distribution of key numeric features in MetroPT
df_metropt.select_dtypes(include=['float64', 'int64']).hist(figsize=(20, 15), bins=20)
plt.suptitle("Distribution of Numeric Features - MetroPT Dataset")
plt.show()
```
